```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df= sns.load_dataset("titanic")
```

```python
df.head()
```

```python
# who adult_male can be droppped
```

```python
df.columns
```

```python
df.info()
```

```python
df.drop(['deck', 'embark_town', 'who', 'alive', 'class', 'adult_male'], axis=1, inplace=True)
```

```python
df.info()
```

```python
df.age.fillna(df['age'].mean(), inplace=True)
```

```python
df.info()
```

```python
df.dropna(subset=['embarked'], inplace=True)# dropping the rows with embarked empty
```

```python
df.info()
```

```python
#now we do label encoding
```

```python
from sklearn.preprocessing import LabelEncoder
le= LabelEncoder()
```

```python
df.head()
```

```python
df['sex']= le.fit_transform(df['sex'])
```

```python
df['embarked']=le.fit_transform(df['embarked'])
```

```python
df=df.astype(int)
```

```python
df.head()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr= LogisticRegression()
```

```python
from sklearn.model_selection import train_test_split
X= df.drop('survived', axis=1)
y=df['survived']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
lr.fit(X_train, y_train)
```

```python
y_pred= lr.predict(X_test)
```

```python
#now for evaluation, we use confusion matrix
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

print(accuracy_score(y_test, y_pred))
print(precision_score(y_test, y_pred))
print(recall_score(y_test, y_pred))
print(f1_score(y_test, y_pred))
```

```python
#for knn we need feature scaling but not for logistic regg
```

```python
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
```

```python
X_train_scaled= sc.fit_transform(X_train)
X_test_scaled= sc.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
k_model= KNeighborsClassifier(n_neighbors=5)
```

```python
k_model.fit(X_train_scaled, y_train)
```

```python
y_pred_knn=k_model.predict(X_test_scaled)
```

```python
print(accuracy_score(y_test, y_pred_knn))
print(precision_score(y_test, y_pred_knn))
print(recall_score(y_test, y_pred_knn))
print(f1_score(y_test, y_pred_knn))
```

```python
# for naive bayes
from sklearn.naive_bayes import GaussianNB

model = GaussianNB()

model.fit(X_train, y_train)

y_pred = model.predict(X_test)
```

```python
print(accuracy_score(y_test,y_pred))
print(f1_score(y_test,y_pred))
```
